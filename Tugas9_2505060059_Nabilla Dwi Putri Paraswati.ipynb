```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Integrasi MongoDB ke PySpark").getOrCreate()
print("Spark berhasil di jalankan")
```

```python
data_mongodb = list(koleksi_menu.find())
print("Jumlah data dari MongoDb:",len(data_mongodb))
```

```python
for menu in data_mongodb:
    print(menu)
```

```python
from pyspark.sql.types import (StructType, StructField, StringType, IntegerType)
skema = StructType([
    StructField("nama_menu", StringType(), True),
    StructField("harga", IntegerType(), True),
    StructField("kategori", StringType(), True),
    StructField("tingkat_kepahitan", IntegerType(), True),
    StructField("pedas_level", IntegerType(), True),
    StructField("stok", IntegerType(), True),
    StructField("status", StringType(), True)])
print("Skema Spark dibuat")
    
```

```python
data_spark = []

for menu in data_mongodb:
    data_spark.append({
        "nama_menu": menu.get("nama_menu"),
        "harga": menu.get("harga"),
        "kategori": menu.get("kategori"),
        "tingkat_kepahitan": menu.get("tingkat_kepahitan"),
        "pedas_level": menu.get("pedas_level"),
        "stok": menu.get("stok"),
        "status": menu.get("status")
    })

df_menu = spark.createDataFrame(data_spark, schema=skema)

print("Data berhasil dikonversi menjadi Spark DataFrame!")
```

```python
df_menu.printSchema()
```

```python
df_menu.show(truncate=False)
```

```python
from pyspark.sql.functions import avg

rata_rata_harga = df_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga")
)

rata_rata_harga.show()
```

Kesimpulan

Berdasarkan praktikum yang telah dilakukan, saya mempelajari penggunaan MongoDB untuk menyimpan dan mengelola data menggunakan operasi CRUD (Create, Read, Update, dan Delete). Saya juga mempelajari cara mengintegrasikan data MongoDB dengan PySpark untuk melakukan pengolahan dan analisis data.

Tantangan yang saya temui adalah struktur dokumen MongoDB yang tidak sama. Pada data menu yang dibuat, kategori Kopi memiliki field tingkat_kepahitan, sedangkan kategori Makanan memiliki field pedas_level. Perbedaan tersebut menjadi tantangan ketika data dikonversi menjadi Spark DataFrame karena Spark membutuhkan skema kolom yang jelas dan sama.

Untuk mengatasinya, saya membuat skema Spark yang mencakup seluruh field dari kedua kategori. Kemudian, saya menggunakan fungsi get() pada Python untuk mengambil setiap field dari dokumen MongoDB. Jika ada dokumen tidak memiliki field tertentu, nilainya akan diisi dengan None dan ditampilkan sebagai null pada Spark DataFrame.

Dari praktikum ini, saya memahami bahwa MongoDB memiliki struktur penyimpanan yang fleksibel, sedangkan Spark DataFrame membutuhkan struktur data yang lebih terorganisasi. Integrasi keduanya dapat membantu proses pengolahan dan analisis data dalam jumlah besar.
